# Per-city descriptive associations

All city-months stay in the calendar grid, with exclusion reasons. Pearson correlations use eligible matched months only. First differences use adjacent calendar months only. Seasonal demeaning is reported only with at least two observations for every calendar month. These are descriptive sensitivity summaries with serial dependence and potential confounding; no p-values, causal effects or pooled cross-city correlation are claimed.

```python
from pathlib import Path
import json
import os
import pandas as pd
from tourism_pipeline import sha256
run_dir = Path(os.environ.get("TOURISM_RUN", "runs/fixture"))
assert json.loads((run_dir / "status.json").read_text())["state"] == "complete"
provenance = json.loads((run_dir / "provenance.json").read_text())
for name, expected in provenance["output_sha256"].items():
    assert sha256(run_dir / name) == expected, name
print("SYNTHETIC SOFTWARE FIXTURE" if provenance["config"]["is_fixture"] else "Observed snapshots; descriptive associations only")
```

```python
import numpy as np
from tourism_pipeline import correlations
combined = pd.read_csv(run_dir / "monthly_combined.csv")
assert not combined.duplicated(["city", "month"]).any()
calculated = correlations(combined)
saved = json.loads((run_dir / "correlations.json").read_text())
for actual, expected in zip(calculated, saved):
    for key, value in expected.items():
        if isinstance(value, float):
            assert np.isclose(actual[key], value, atol=1e-12), key
        else:
            assert actual[key] == value, key
pd.DataFrame(saved)
```
