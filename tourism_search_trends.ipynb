# Relative search interest and term coverage

This notebook reads the same saved run as the 311 analysis. Google Trends scores are normalized relative interest, not visitor counts or absolute search volume. The index is an equal-weight mean of the fixed term basket; each city-month requires every term to be usable. Missing, partial and censored terms remain visible. Cross-city index levels are not a tourism-volume comparison.

```python
from pathlib import Path
import json
import os
import pandas as pd
from tourism_pipeline import sha256
run_dir = Path(os.environ.get("TOURISM_RUN", "runs/fixture"))
assert json.loads((run_dir / "status.json").read_text())["state"] == "complete"
provenance = json.loads((run_dir / "provenance.json").read_text())
for name, expected in provenance["output_sha256"].items():
    assert sha256(run_dir / name) == expected, name
print("SYNTHETIC SOFTWARE FIXTURE" if provenance["config"]["is_fixture"] else "Observed snapshots; descriptive associations only")
```

```python
trends = pd.read_csv(run_dir / "monthly_trends.csv")
assert not trends.duplicated(["city", "month"]).any()
trends.groupby("city")[["trends_complete", "n_partial_terms", "n_censored_terms"]].sum()
```
